## New Metrics

### Metric 1: Total Goals

```python
df_matches_historical['total_goals'] = df_matches_historical['home_score'] + df_matches_historical['away_score']
df_matches_modern['total_goals'] = df_matches_modern['home_score'] + df_matches_modern['away_score']
```

```python
df_matches_historical['total_goals'].head()
```

### Metric 2: Match Winner

```python
def get_winner(row):
    if row['home_score'] > row['away_score']:
        return 'Home Win'
    elif row['away_score'] > row['home_score']:
        return 'Away Win'
    else:
        return 'Draw'

df_matches_historical['match_winner'] = df_matches_historical.apply(get_winner, axis=1)
df_matches_modern['match_winner'] = df_matches_modern.apply(get_winner, axis=1)
```

```python
# kolom  untuk Q1, Q2, dan Q3
columns = ['year', 'home_team', 'away_team', 'host', 'total_goals', 'match_winner']

# 2. Ambil subset kolom dari kedua DataFrame
df_historical = df_matches_historical[columns]
df_modern = df_matches_modern[columns] # Kolom xG akan otomatis terabaikan

# concat untuk analisis 1930-2022
df_full_analysis = pd.concat([df_historical, df_modern], ignore_index=True)
```

# Agregasi Data untuk tiap Pertanyaan Analitis Spesifik

## Q1: Evolusi Gol dari 1930 - 2014

```python
# agregasi
# df_q1 = df_matches_historical.groupby('year')['total_goals'].mean().reset_index()
df_q1 = df_full_analysis.groupby('year')['total_goals'].mean().reset_index()

df_q1
# HASIL: df_q1 adalah DataFrame kecil (20 baris) yang siap untuk st.line_chart()
```

## Q2: Gol vs Popularitas

```python
# (Asumsikan df_world_cup sudah bersih dan punya kolom 'year' dan 'attendance_avg')
df_world_cup_simple = df_worldcup[['year', 'attendance_avg']]
# merge
df_q2 = pd.merge(df_q1, df_world_cup_simple, on='year')

df_q2
# HASIL: df_q2 siap untuk st.scatter_chart(x='total_goals', y='attendance_avg')
```

## Q3: Home Advantage

```python
# 1. Hitung Host Win Percentage
host_matches = df_full_analysis[df_full_analysis['home_team'] == df_full_analysis['host']]
host_wins = host_matches[host_matches['match_winner'] == 'Home Win'].shape[0]
host_win_perc = host_wins / host_matches.shape[0]

# 2. Hitung Baseline Win Percentage (Semua laga 'home' di data historis)
total_matches_hist = df_full_analysis.shape[0]
home_wins_hist = df_full_analysis[df_full_analysis['match_winner'] == 'Home Win'].shape[0]
baseline_win_perc_hist = home_wins_hist / total_matches_hist

# 3. Buat DataFrame untuk di-plot
df_q3 = pd.DataFrame({
    'Category': ['Tim Tuan Rumah (Host)', 'Baseline (Semua Laga)'],
    'Win Percentage': [host_win_perc, baseline_win_perc_hist]
})

df_q3
# HASIL: df_q3 siap untuk st.bar_chart()
```

## Agregasi Bonus

```python
# agregasi per tim (saat bermain sebagai 'home_team')
df_matches_modern_agg = df_matches_modern.groupby('home_team').agg({
    'home_score': 'sum',
    'home_xg': 'sum'
}).reset_index().rename(columns={'home_team':'team'})

# metrik baru performa vs xG
df_matches_modern_agg['performance_vs_xg'] = df_matches_modern_agg['home_score'] - df_matches_modern_agg['home_xg']

# HASIL: df_matches_modern_agg siap untuk di-plot (misal: bar chart perbandingan score vs xG)
```

```python
# 1. Agregasi statistik saat tim bermain 'home'
df_home_stats = df_matches_modern.groupby('home_team').agg(
    home_score_total=('home_score', 'sum'),
    home_xg_total=('home_xg', 'sum')
).reset_index().rename(columns={'home_team': 'team'})

# 2. Agregasi statistik saat tim bermain 'away'
df_away_stats = df_matches_modern.groupby('away_team').agg(
    away_score_total=('away_score', 'sum'),
    away_xg_total=('away_xg', 'sum')
).reset_index().rename(columns={'away_team': 'team'})

# 3. Gabungkan (Merge) kedua data berdasarkan nama tim
# Kita gunakan 'outer' merge untuk memastikan tidak ada tim yang hilang
df_matches_modern_agg = pd.merge(df_home_stats, df_away_stats, on='team', how='outer')

# 4. Isi nilai NaN (jika ada tim yg hanya main home/away) dengan 0
df_matches_modern_agg = df_matches_modern_agg.fillna(0)

# 5. BUAT METRIK TOTAL (MENGGABUNGKAN)
# (Sambil tetap mempertahankan data terpisah)
df_matches_modern_agg['total_goals'] = df_matches_modern_agg['home_score_total'] + df_matches_modern_agg['away_score_total']
df_matches_modern_agg['total_xg'] = df_matches_modern_agg['home_xg_total'] + df_matches_modern_agg['away_xg_total']

# 6. Metrik Performa Total
df_matches_modern_agg['performance_vs_xg'] = df_matches_modern_agg['total_goals'] - df_matches_modern_agg['total_xg']
```
